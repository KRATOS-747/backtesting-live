### 1. Executive Summary & Quantitative Thesis

> **The Core Distinction**:
> * **The Raw Edge**: The structural probability tailwind provided by market rules or participant imbalances (here, the Variance Risk Premium: $	ext{IV} > 	ext{RV}$ and positive Theta decay $\Theta > 0$).
> * **The Alpha**: The quantitative feature engineering and risk conditioning that filters false signals, suppresses friction churn, and allows the desk to retain profits after real-world exchange taxes, GST, and slippage.

In retail trading and naive backtesting, selling ATM straddles and rolling to the new ATM whenever spot moves $\pm 40$ points is widely considered a baseline strategy. 
However, when subjected to **rigorous Indian statutory frictions** (STT, exchange turnover charges, GST 18%, stamp duty, and brokerage), the flat 40pt rolling straddle executes **258 rolls in 23 days**, paying **₹35,118.41 in fees** and finishing with a net **loss of -₹3,453.41**.

By engineering a **5-Minute Spot Rate-of-Change (ROC) Elastic Boundary**, we:
* Cut turnover by **37%** (from 258 down to 163 rolls).
* Save **₹11,730 in cold cash** from being handed to brokers/exchanges.
* Boost Gross P&L by **+38%** (from ₹31.6k to ₹43.6k) by avoiding entries in front of momentum impulses.
* Swing Net Realized P&L from **-₹3,453.41 to +₹20,221.57** (Win rate jumping from 43.5% to 60.9%).
