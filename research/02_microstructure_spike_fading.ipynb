### 6. The Double-Leg Straddle Penalty Lesson

During our research, we also evaluated selling a 2-leg ATM Straddle rather than a single naked leg on the same spike triggers:
* **Gross P&L expanded**: +₹55,507.50 (+23% more gross premium decay).
* **Friction nearly doubled**: -₹64,530.21 (4 order executions, ₹80 brokerage, and 2.0 pt spread crossing).
* **Net Realized Outcome**: **-₹9,022.71 (Net Loss)**.

**Key Finding**: In fast event-driven intraday scalps, every additional leg is a severe friction liability. Straddles must be reserved for multi-hour macro Theta harvesting (Alpha 1), while single-leg execution remains optimal for event fades (Alpha 2).
